```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
print("Loading the dataset..")
df=pd.read_csv('jobs_dataset.csv')
print("Dataset loaded succefully! Shape:",df.shape)
```

```python
df.head()
```

```python

# Data Cleaning

# Drop rows with missing values in the salary column

df = df.dropna(subset=["salary"]).copy()

# Extract numerical salary values

df["salary_num"] = (
df["salary"]
.astype(str)
.str.replace(",", "", regex=False)
.str.extract(r"(\d+(?:.\d+)?)", expand=False)
)

# Convert salary values to numeric data type

df["salary_num"] = pd.to_numeric(df["salary_num"], errors="coerce")

# Drop rows where numeric extraction failed

df = df.dropna(subset=["salary_num"]).copy()

# Display the results

display(df[["salary", "salary_num"]].head(10))

print("Dataset shape after cleaning:", df.shape)


```

```python
#Definefeatures X and y

X = df[["company", "rating", "location", "positionName"]].copy()

# Define target variable

y = df["salary_num"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#Handle Missing Values
#Fill missing rating values using the median
X["rating"] = X["rating"].fillna(X["rating"].median())

print("Missing ratings:", X["rating"].isna().sum())
```

```python
#One Hot Encoding
X_encoded = pd.get_dummies(
X,
columns=["company", "location", "positionName"],
dtype=int
)

display(X_encoded.head())

print("Encoded X shape:", X_encoded.shape)
```

```python
#Build and Train the Model
model = DecisionTreeRegressor(random_state=42)

# Train the model

model.fit(X_encoded, y)

print("Model trained successfully!")
```

```python
#Predict Salaries
y_pred = model.predict(X_encoded)

comparison = pd.DataFrame({
"Actual Salary": y,
"Predicted Salary": y_pred
})

display(comparison.head(10))
```

```python
#Visualize the Decision Tree
plt.figure(figsize=(20, 8))

plot_tree(
model,
feature_names=X_encoded.columns,
filled=True,
rounded=True,
max_depth=3,
fontsize=8
)

plt.title("Decision Tree Regressor")
plt.show()
```
